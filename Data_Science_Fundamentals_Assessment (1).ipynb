## 1. Data Science Fundamentals

### 1.1 The Data Science Lifecycle
| Stage | Description | Example (customer churn) |
|---|---|---|
| 1. Problem definition | Turn a business question into a measurable goal | Reduce monthly churn by 10% |
| 2. Data collection | Gather data from databases, APIs, files, sensors | CRM records, usage logs |
| 3. Data cleaning | Fix missing values, duplicates, wrong types, outliers | Impute missing tenure |
| 4. Exploratory analysis (EDA) | Summaries, distributions, relationships | Churn rate by plan type |
| 5. Modelling | Build statistical / ML models | Logistic regression |
| 6. Evaluation | Measure performance with suitable metrics | Precision, recall, AUC |
| 7. Deployment & communication | Ship the model, report findings | Dashboard, weekly retention list |
| 8. Monitoring | Track drift and re-train | Alert if accuracy drops |

The lifecycle is **iterative**: findings in EDA often send you back to data collection or cleaning.

### 1.2 Types of Analysis
| Type | Question | Example | Typical techniques |
|---|---|---|---|
| Descriptive | What happened? | Monthly sales totals | Mean, counts, charts |
| Diagnostic | Why did it happen? | Why did sales dip in March? | Drill-down, correlation |
| Predictive | What will happen? | Next-quarter demand | Regression, classification |
| Prescriptive | What should we do? | Optimal price / stock level | Optimisation, simulation |

### 2.4 Functions

```python
def describe(numbers, round_to=2):
    """Return a dict with the count, mean, min and max of `numbers`.

    Parameters
    ----------
    numbers : list[float]
    round_to : int, number of decimal places for the mean
    """
    if not numbers:
        raise ValueError("numbers must not be empty")
    return {
        "count": len(numbers),
        "mean": round(sum(numbers) / len(numbers), round_to),
        "min": min(numbers),
        "max": max(numbers),
    }

print(describe([4, 8, 15, 16, 23, 42]))

square = lambda x: x ** 2           # lambda (anonymous function)
print(list(map(square, [1, 2, 3])))
```

## 3. Statistical Foundations
We use a small example: exam scores for 15 students (one deliberate outlier).

```python
import numpy as np
import pandas as pd
from statistics import mean, median, mode, pvariance, variance, stdev

scores = [55, 62, 65, 68, 70, 70, 72, 75, 78, 80, 82, 85, 88, 90, 150]  # 150 = outlier
```

### 3.1 Central Tendency and Spread

```python
print(f"Mean              : {mean(scores):.2f}")
print(f"Median            : {median(scores):.2f}")
print(f"Mode              : {mode(scores)}")
print(f"Range             : {max(scores) - min(scores)}")
print(f"Variance (sample) : {variance(scores):.2f}")
print(f"Std dev (sample)  : {stdev(scores):.2f}")
print(f"Variance (pop.)   : {pvariance(scores):.2f}")

# Manual check of the sample variance formula  s^2 = sum((x - mean)^2) / (n - 1)
m = mean(scores)
manual = sum((x - m) ** 2 for x in scores) / (len(scores) - 1)
print("Manual variance matches:", np.isclose(manual, variance(scores)))
```

**Interpretation:** the mean (≈ 79.3) is pulled above the median (75) by the outlier at 150, so the **median is the more robust** measure here. Use the *sample* formula (n−1) when data is a sample of a larger population.

### 3.2 Percentiles, Quartiles and Outlier Detection (IQR rule)

```python
q1, q2, q3 = np.percentile(scores, [25, 50, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print(f"Q1={q1}, Median={q2}, Q3={q3}, IQR={iqr}")
print(f"Outlier fences: [{lower}, {upper}]")
print("Outliers:", [x for x in scores if x < lower or x > upper])
print("90th percentile:", np.percentile(scores, 90))
```

### 3.3 Probability Basics

```python
# Classical probability: P(event) = favourable outcomes / total outcomes
p_even_die = 3 / 6
print("P(even on a fair die)           =", p_even_die)

# Independent events: P(A and B) = P(A) * P(B)
print("P(two heads in two coin flips)  =", 0.5 * 0.5)

# Complement: P(not A) = 1 - P(A)
print("P(at least one 6 in 3 rolls)    =", round(1 - (5 / 6) ** 3, 4))

# Conditional probability: P(A|B) = P(A and B) / P(B)
# 60% of customers use mobile; 30% use mobile AND buy.
p_buy_given_mobile = 0.30 / 0.60
print("P(buy | mobile)                 =", p_buy_given_mobile)

# Simulation check of the law of large numbers
rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, size=100_000)
print("Simulated P(even)               =", round((rolls % 2 == 0).mean(), 4))
```

### 3.4 Correlation

```python
rng = np.random.default_rng(42)
hours = rng.uniform(1, 10, 50)
marks = 40 + 5 * hours + rng.normal(0, 5, 50)       # strong positive relationship
noise = rng.normal(0, 1, 50)                        # unrelated variable

df_corr = pd.DataFrame({"hours_studied": hours, "marks": marks, "noise": noise})
print(df_corr.corr().round(3))
```

**Interpretation:** Pearson *r* ranges from −1 to +1. `hours_studied` vs `marks` shows a strong positive correlation, while `noise` is close to 0. **Correlation ≠ causation**: a high *r* only shows association; confounders or coincidence can produce it.

## 4. Data Types and Structures

| Type | Description | Examples | Typical handling |
|---|---|---|---|
| Numerical – continuous | Measurable, any value in range | height, price | Mean, std, scaling |
| Numerical – discrete | Countable integers | number of orders | Counts, Poisson-type models |
| Categorical (nominal) | Labels with no order | city, colour | One-hot encoding |
| Ordinal | Ordered categories | Low/Med/High | Ordered encoding (preserve rank) |
| Datetime | Timestamps | order date | Parse, extract parts, resample |
| Text | Free-form strings | reviews | Clean, tokenise, vectorise |
| Boolean | True/False | is_active | 0/1 |

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106],
    "age":         [25, 34, None, 45, 29, 52],                 # numerical
    "city":        ["Pune", "Delhi", "Pune", "Mumbai", "Delhi", "Pune"],   # nominal
    "satisfaction":["Low", "High", "Medium", "High", "Low", "Medium"],     # ordinal
    "signup_date": ["2024-01-15", "2024-02-20", "2024-03-05",
                    "2024-03-28", "2024-04-10", "2024-05-01"],  # datetime (as text)
    "review":      ["Great service!", "Too slow...", "Okay, not bad",
                    "LOVED it", "Poor support", "fine"],
})
print(df.dtypes, "\n")

# Convert to appropriate types
df["city"] = df["city"].astype("category")
df["satisfaction"] = pd.Categorical(df["satisfaction"],
                                    categories=["Low", "Medium", "High"], ordered=True)
df["signup_date"] = pd.to_datetime(df["signup_date"])
print(df.dtypes)
```

```python
# Handle each type appropriately
df["satisfaction_code"] = df["satisfaction"].cat.codes            # ordinal -> ranks 0,1,2
city_dummies = pd.get_dummies(df["city"], prefix="city")          # nominal -> one-hot
df["signup_month"] = df["signup_date"].dt.month_name()            # datetime -> feature
df["review_clean"] = (df["review"].str.lower()
                      .str.replace(r"[^a-z ]", "", regex=True).str.strip())  # text cleaning
df["review_word_count"] = df["review_clean"].str.split().str.len()

print(pd.concat([df[["customer_id", "satisfaction", "satisfaction_code",
                     "signup_month", "review_clean", "review_word_count"]],
                 city_dummies], axis=1).to_string(index=False))
print("\nIs 'High' > 'Low'?", df.loc[1, "satisfaction"] > df.loc[0, "satisfaction"])
```

## 5. Data Science Best Practices

| Practice | Poor | Better |
|---|---|---|
| Meaningful names | `d`, `x1`, `tmp` | `sales_df`, `avg_order_value` |
| Small, documented functions | one 200-line cell | functions with docstrings |
| Don't modify raw data | overwrite source file | work on `.copy()`, save cleaned output separately |
| Reproducibility | random results each run | fixed seeds, pinned library versions, notebook that runs top-to-bottom |
| Constants not magic numbers | `if x > 0.37` | `CHURN_THRESHOLD = 0.37` |
| Validate data | assume it is clean | assertions / checks on shape, ranges, nulls |
| Responsible interpretation | "X causes Y" from a correlation | state limits, sample size, bias, uncertainty |

```python
SEED = 42
MIN_VALID_AGE, MAX_VALID_AGE = 0, 120

def clean_customers(raw_df):
    """Return a cleaned copy of the customer table (raw data is not modified).

    Steps: drop duplicate ids, impute missing age with the median, validate range.
    """
    clean_df = raw_df.copy()
    clean_df = clean_df.drop_duplicates(subset="customer_id")
    median_age = clean_df["age"].median()
    clean_df["age"] = clean_df["age"].fillna(median_age)
    assert clean_df["age"].between(MIN_VALID_AGE, MAX_VALID_AGE).all(), "Age out of range"
    return clean_df

cleaned = clean_customers(df[["customer_id", "age"]])
print("Raw missing ages    :", df["age"].isna().sum())
print("Cleaned missing ages:", cleaned["age"].isna().sum())
```

**Responsible interpretation checklist:** Is the sample representative? Could a confounder explain the pattern? Are outliers distorting the summary? Is the result practically (not just statistically) meaningful? Have I reported uncertainty and limitations honestly? Does the data contain personal information that must be protected?

**Task:** Write a function that takes a sentence and returns the word frequencies (ignoring case and punctuation), then report the top 3 words.

```python
import string
from collections import Counter

def word_frequencies(text):
    """Return a Counter of lower-cased words with punctuation removed."""
    cleaned = text.lower().translate(str.maketrans("", "", string.punctuation))
    return Counter(cleaned.split())

text = "Data is the new oil. Data without analysis is just data, and analysis needs data!"
print(word_frequencies(text).most_common(3))
```

### Assignment 2 – Statistics
**Task:** For daily sales values, compute the summary statistics, detect outliers with the IQR rule, and compare the mean with and without the outliers.

```python
sales = pd.Series([220, 240, 235, 250, 245, 230, 260, 255, 238, 900])  # 900 = promo spike
q1, q3 = sales.quantile([0.25, 0.75])
iqr = q3 - q1
mask = sales.between(q1 - 1.5 * iqr, q3 + 1.5 * iqr)

print(sales.describe().round(2))
print("\nOutliers:", sales[~mask].tolist())
print(f"Mean with outliers   : {sales.mean():.1f}")
print(f"Mean without outliers: {sales[mask].mean():.1f}")
```

**Conclusion:** one promotional spike lifts the mean from about 241 to 307, well above any typical day. Report the median or investigate the spike separately rather than silently deleting it.

**Task:** Using a small orders dataset: fix types, handle missing values, filter, group, and summarise.

```python
rng = np.random.default_rng(SEED)
orders = pd.DataFrame({
    "order_id": range(1, 13),
    "order_date": pd.date_range("2024-01-01", periods=12, freq="10D").astype(str),
    "region": rng.choice(["North", "South", "East", "West"], 12),
    "category": rng.choice(["Electronics", "Clothing", "Grocery"], 12),
    "amount": rng.integers(200, 2000, 12).astype(float),
    "rating": rng.choice([1, 2, 3, 4, 5, None], 12),
})
orders.loc[2, "amount"] = np.nan          # inject a missing value

# 1. Fix types
orders["order_date"] = pd.to_datetime(orders["order_date"])
orders["region"] = orders["region"].astype("category")

# 2. Missing values
print("Missing per column:\n", orders.isna().sum(), "\n")
orders["amount"] = orders["amount"].fillna(orders["amount"].median())

# 3. Feature engineering + filtering
orders["month"] = orders["order_date"].dt.to_period("M")
high_value = orders[orders["amount"] > 1000]
print("High-value orders:", len(high_value), "of", len(orders), "\n")

# 4. Group and aggregate
summary = (orders.groupby("category", observed=True)["amount"]
           .agg(orders="count", total="sum", average="mean")
           .round(1).sort_values("total", ascending=False))
print(summary, "\n")
print(orders.groupby("month")["amount"].sum().round(0))
```

### Assignment 4 – Basic analysis (EDA) with interpretation
**Task:** Explore the relationship between advertising spend and revenue.

```python
rng = np.random.default_rng(SEED)
ad_spend = rng.uniform(10, 100, 60)                       # in thousands
revenue = 50 + 3.2 * ad_spend + rng.normal(0, 25, 60)     # in thousands
eda = pd.DataFrame({"ad_spend": ad_spend, "revenue": revenue})

print(eda.describe().round(2), "\n")
r = eda["ad_spend"].corr(eda["revenue"])
print(f"Pearson r = {r:.3f}")

# Simple linear fit: revenue = slope * ad_spend + intercept
slope, intercept = np.polyfit(eda["ad_spend"], eda["revenue"], 1)
r2 = r ** 2
print(f"Fit: revenue = {slope:.2f} * ad_spend + {intercept:.1f}   (R^2 = {r2:.2f})")
```

**Interpretation:** there is a strong positive association (*r* ≈ 0.97, R² ≈ 0.95): each extra 1k of ad spend goes with roughly 3k more revenue in this (synthetic) data. This is **association, not proven causation**; seasonality, pricing, or market size could also drive both variables. The sample is small and the linear fit should only be applied within the observed spend range (10k–100k).
